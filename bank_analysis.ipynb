```python
import yfinance as yf
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from IPython.display import display, HTML
```

```python
#Defining Assets to be compared
tickers = {
    "yes_bank": "YESBANK.NS",
    "indusind_bank": "INDUSINDBK.NS",
    "axis_bank": "AXISBANK.NS",
    "pnb": "PNB.NS",
    "bank_nifty": "^NSEBANK"
}

start_date = "2005-01-01"
end_date = None  # None = up to today
```

```python
# --- Pull closing prices for each ticker ---
price_data = {}
for name, ticker in tickers.items():
    hist = yf.download(ticker, start=start_date, end=end_date, progress=False)
    price_data[name] = hist["Close"]
    print(f"{name} ({ticker}): {len(hist)} rows, {hist.index.min().date()} to {hist.index.max().date()}")
```
